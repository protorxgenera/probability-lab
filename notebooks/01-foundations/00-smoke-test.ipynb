```python
import numpy as np
rng = np.random.default_rng(0)

# PREDICT first: P(sick | positive)? Write your guess here: ____
N = 1_000_000
sick = rng.random(N) < 0.01                 # 1% prevalence
pos = np.where(sick, rng.random(N) < 0.95,  # sensitivity
                     rng.random(N) < 0.05)  # false positive rate
print("simulated:", sick[pos].mean())

# DERIVE: Bayes by hand, then check
exact = 0.95*0.01 / (0.95*0.01 + 0.05*0.99)
print("exact:   ", exact)
```
