## 4.1 Defining and calling functions

```python
def name(param1, param2):
    """Docstring: one line saying what the function does."""
    ...body...
    return result
```

- `return` sends a value **back to the caller** and ends the function.
- A function without `return` returns `None`.
- **`print` is not `return`.** `print` only shows text on screen; the caller gets nothing it can compute with.

````python
def mean(xs):
    """Arithmetic mean of a non-empty list."""
    return sum(xs) / len(xs)

def mean_printing(xs):
    print(sum(xs) / len(xs))     # shows the value but returns None

m1 = mean([1, 2, 3])
m2 = mean_printing([1, 2, 3])
print("m1 =", m1, "| m2 =", m2)  # m2 is None!
````

> **Note on the MLMA lab:** its `greet` example ends with `return print(...)`. That returns `None`
> (the return value of `print`) and prints the line twice. Now you know why.

## 4.2 Default and keyword arguments

- **Default values** make a parameter optional: `def f(x, p=0.5)`.
- **Keyword arguments** name the parameter at the call site, so order doesn't matter: `f(p=0.2, x=3)`.
- Positional arguments come first, then keyword arguments.
- **Never use a mutable default** like `def f(xs=[])`. Use `None` and create the list inside.

````python
def greet(name="User", age=20):
    return f"Hello, {name}. You are {age} years old."

print(greet())
print(greet("Ana"))
print(greet(age=25, name="John"))
````

## 4.3 Returning several values

Return a tuple and unpack it at the call site, exactly like `u, s, vh = svd(matrix)` in the lab.

````python
def min_max(xs):
    return min(xs), max(xs)

lo, hi = min_max([4, 1, 7, 3])
print(lo, hi)
````

## 4.4 `lambda`: tiny anonymous functions

`lambda x: x * x` is a function with one expression and no name. They're mostly used as **arguments**
to other functions, e.g. `sorted(..., key=lambda ...)` or `minimize(lambda x: ..., x0=0)`.

````python
square = lambda x: x * x
print(square(5))

people = [("Ana", 31), ("Bogdan", 24), ("Cris", 28)]
print(sorted(people, key=lambda person: person[1]))   # sort by age
````

## 4.5 Functions are values

You can store a function in a variable, put it in a list, and pass it into another function.
This is how numerical libraries work: you give them *your* function and they evaluate it many times.

````python
def apply_twice(f, x):
    return f(f(x))

print(apply_twice(square, 3))           # (3^2)^2 = 81
print(list(map(square, [1, 2, 3])))     # map applies f to every element
````

## 4.6 Scope

Variables created inside a function are **local**: they disappear when the function returns and don't
affect variables with the same name outside. Functions should get inputs via parameters and give outputs via `return`.

````python
total = 100
def add_up(xs):
    total = 0           # a different, local 'total'
    for x in xs:
        total += x
    return total

print(add_up([1, 2, 3]), total)   # 6 100
````

## 4.7 `*args` and `**kwargs` (reading them, mostly)

You'll see these in library signatures: `*args` collects extra positional arguments into a tuple and `**kwargs`
collects extra keyword arguments into a dict. A common use is forwarding options to another function.

````python
def describe(*args, **kwargs):
    print("args:", args, "kwargs:", kwargs)

describe(1, 2, color="red", linestyle="--")
````
