## 6.1 Creating arrays

An `ndarray` is a grid of numbers that all have **one** type. It's much faster than a list for math.

| function | gives | example |
|---|---|---|
| `np.array(list)` | from a (nested) list | `np.array([[1, 2], [3, 4]])` |
| `np.arange(start, stop, step)` | like `range`, can use floats | `np.arange(0, 10, 2)` |
| `np.linspace(a, b, n)` | `n` evenly spaced points **including** `b` | `np.linspace(0, 1, 5)` |
| `np.zeros(shape)`, `np.ones(shape)` | filled arrays | `np.zeros((2, 3))` |
| `np.full(shape, v)` | filled with `v` | `np.full(3, 0.5)` |
| `np.eye(n)`, `np.diag(v)` | identity / diagonal matrix | `np.diag([1, 2, 3])` |

Use `arange` for integer steps and `linspace` for "n points on an interval", e.g. the x-axis of a plot.

````python
a = np.array([1, 2, 3, 4, 5])
print(a, np.arange(0, 10, 2), np.linspace(0, 1, 5))
print(np.zeros((2, 3)))
print(np.diag([1, 2, 3]))
````

## 6.2 Shape, dtype and dimensions

| attribute | meaning |
|---|---|
| `a.shape` | tuple of sizes per dimension: `(rows, cols)` for a matrix |
| `a.ndim` | number of dimensions |
| `a.size` | total number of elements |
| `a.dtype` | element type: `int64`, `float64`, `bool`, … |

**Row vs. column vector.** `np.array([1, 2, 3])` has shape `(3,)`: it's 1-D, neither a row nor a column.
`np.array([[1], [2], [3]])` has shape `(3, 1)`, a real column (2-D). Most bugs in matrix code are shape bugs,
so **print `.shape` often**.

````python
matrix = np.array([[1, 2], [3, 4], [5, 6]])
row_vector = np.array([1, 2, 3])
column_vector = np.array([[1], [2], [3]])
print(matrix.shape, matrix.ndim, matrix.size, matrix.dtype)
print(row_vector.shape, column_vector.shape)
print(matrix.astype(float).dtype)       # type casting
````

## 6.3 Indexing and slicing

1-D works like lists. In 2-D use **one bracket with a comma**: `M[row, col]`.
A `:` means "everything along this axis".

```
M[1, 2]     element in row 1, column 2
M[1, :]     row 1 (also M[1])
M[:, 0]     column 0
M[:2, 1:]   rows 0-1, columns 1-end
```

**Slices are views**, not copies: changing a slice changes the original array. Use `.copy()` when you need independence.

````python
M = np.arange(12).reshape(3, 4)
print(M)
print(M[1, 2], M[1, :], M[:, 0], M[-1, -1])
print(M[:2, 1:])

view = M[0, :]
view[0] = 100
print(M[0])          # the original changed too!
````

## 6.4 Vectorized operations

Arithmetic on arrays is **element-wise**, with no loop needed. NumPy also has element-wise versions of
the math functions: `np.sqrt`, `np.exp`, `np.log`, `np.sin`, `np.abs`, …

Watch out: `*` is element-wise. Matrix multiplication is `@` (Chapter 7).

````python
x = np.array([1.0, 2.0, 3.0])
y = np.array([10.0, 20.0, 30.0])
print(x + y, x * y, x ** 2, 5 * x, 1 / x)
print(np.exp(x), np.sqrt(x))

# Summaries are methods (or functions)
print(x.sum(), x.mean(), x.std(), x.min(), x.max(), np.median(x))
````

## 6.5 Boolean masks

A comparison on an array gives an array of `True`/`False`, called a **mask**. You can:
- index with it to **filter**: `a[a > 2]`
- `sum` it to **count** (True counts as 1): `(a > 2).sum()`
- `mean` it to get a **proportion**: `(a > 2).mean()`. That's an empirical probability!
- combine masks with `&` (and), `|` (or), `~` (not). **Use parentheses**: `(a > 1) & (a < 4)`.

`np.where(cond, x, y)` picks from `x` where `cond` is true and from `y` elsewhere.
`np.where(cond)` with one argument returns the **indices** where `cond` is true.

````python
a = np.array([3, 8, 1, 9, 4, 7])
print(a > 4)
print(a[a > 4], (a > 4).sum(), (a > 4).mean())
print(a[(a > 2) & (a < 8)])
print(np.where(a > 4, "big", "small"))
print(np.where(a > 4))
````

## 6.6 Fancy indexing

Index with a **list or array of integers** to pick several elements at once, in any order and with repeats allowed.
`np.take(a, idx)` does the same thing.

````python
faces = np.array([10, 20, 30, 40])
print(faces[[0, 2, 2, 3]], np.take(faces, [0, 2]))
````

### Exercise 6.4: Replace a loop with vectorized code
The loop below computes the standard normal density at many points. Write `phi_vec`, the same values computed
**without a loop** (use `np.exp`, `np.sqrt`, `np.pi` directly on the array `xs`).

````python
import math
xs = np.linspace(-3, 3, 7)

phi_loop = []
for x in xs:
    phi_loop.append(math.exp(-x ** 2 / 2) / math.sqrt(2 * math.pi))

phi_vec = np.exp(-xs ** 2 / 2) / np.sqrt(2 * np.pi)
````

````python
assert isinstance(phi_vec, np.ndarray) and phi_vec.shape == xs.shape
assert np.allclose(phi_vec, phi_loop)
print("Exercise 6.4 passed")
````

### Exercise 6.5: Summary statistics

### Exercise 6.6: Boolean masks: filter, count, proportion
For `temps` (daily temperatures):
- `hot_days`: the temperatures above 30
- `n_hot`: how many there are (sum a mask)
- `frac_mild`: the **fraction** of days with 15 ≤ temp ≤ 25 (mean of a combined mask)

### Exercise 6.7: Masks on a matrix
`grades` has one row per student and one column per exam.
- `passed`: a boolean matrix, True where a grade is ≥ 5
- `n_passed_per_student`: how many exams each student passed (sum along columns: `.sum(axis=1)`)
- `students_all_passed`: indices of students who passed every exam (`np.where` with one argument, then `[0]`)

<details><summary>Hint</summary>

`np.where(cond)` returns a *tuple* of index arrays, one per dimension. For a 1-D condition take element `[0]`.
`axis=1` means "collapse the columns, giving one value per row". Chapter 7 goes deeper.

````python
grades = np.array([[7, 4, 9],
                   [5, 6, 8],
                   [3, 2, 6],
                   [9, 9, 10]])
passed = grades >= 5
n_passed_per_student = passed.sum(axis=1)
students_all_passed = np.where(n_passed_per_student == grades.shape[1])[0]
````

### Exercise 6.10: dtype and casting
- `as_float`: `counts` converted to `float64` with `.astype`
- `as_int`: `measurements` converted to integers (note: `.astype(int)` truncates, it doesn't round)

### Exercise 6.11: A grid of points with `meshgrid`
To evaluate a function of two variables on a grid (for a contour plot or a 2-D density), use
`X, Y = np.meshgrid(xs, ys)`. Compute `Z`, the values of $f(x, y) = x^2 + y^2$ on the grid below.
Then set `min_location` to the `(x, y)` tuple of grid coordinates where `Z` is smallest.

<details><summary>Hint</summary>

`np.argmin(Z)` gives a *flat* index. `np.unravel_index(flat_index, Z.shape)` turns it into `(row, col)`.
Rows correspond to `ys`, columns to `xs`.
</details>

````python
xs = np.linspace(-2, 2, 5)
ys = np.linspace(-1, 3, 5)
X, Y = np.meshgrid(xs, ys)
Z = X ** 2 + Y ** 2
row, col = np.unravel_index(np.argmin(Z), Z.shape)
min_location = (xs[col], ys[row])
````

````python
assert Z.shape == (5, 5) and np.isclose(Z[0, 0], 4 + 1)
assert tuple(float(c) for c in min_location) == (0.0, 0.0)
print("Exercise 6.11 passed")
````

### Exercise 6.12: Empirical probability from an array of outcomes
`rolls` holds 20 die rolls. Compute **without loops**:
- `p_six`: the fraction of sixes
- `p_even`: the fraction of even rolls
- `freqs`: an array of length 6 with the relative frequency of each face 1..6.
  Try `np.bincount(rolls, minlength=7)[1:] / rolls.size`, and work out why the `[1:]` is there.

````python
rolls = np.array([3, 6, 1, 6, 2, 5, 4, 6, 1, 3, 2, 6, 5, 4, 3, 1, 6, 2, 2, 4])
p_six = (rolls == 6).mean()
p_even = (rolls % 2 == 0).mean()
freqs = np.bincount(rolls, minlength=7)[1:] / rolls.size
````

````python
assert np.isclose(p_six, 0.25) and np.isclose(p_even, 0.6)
assert np.allclose(freqs, [0.15, 0.2, 0.15, 0.15, 0.1, 0.25]) and np.isclose(freqs.sum(), 1)
print("Exercise 6.12 passed")
````
