## 1.1 Variables and the basic types

A **variable** is a name that points to a value. `=` means *assign*, not *equals*.
Python has four basic number-ish types:

| type | example | meaning |
|---|---|---|
| `int` | `10`, `-3`, `1_000_000` | whole numbers (any size) |
| `float` | `20.5`, `1e-3`, `0.1` | real numbers (64-bit floating point) |
| `bool` | `True`, `False` | truth values |
| `complex` | `1 + 3j` | complex numbers (`j` is the imaginary unit) |

`type(x)` tells you the type of a value. `int(...)`, `float(...)`, `str(...)` and `bool(...)` **convert** (cast) between types.

```python
age = 25
height = 1.82
is_student = True
z = 1 + 3j

print(type(age), type(height), type(is_student), type(z))
print(float(age))      # 25.0
print(int(3.99))       # 3  -> int() truncates toward zero, it does NOT round
print(round(3.99))     # 4
print(int("42") + 1)   # strings that look like numbers can be converted
print(bool(0), bool(7), bool(""), bool("hi"))  # zero/empty is False, everything else is True
```

## 1.2 Arithmetic

| operator | meaning | example | result |
|---|---|---|---|
| `+ - *` | the usual | `7 * 2` | `14` |
| `/` | true division (always a float) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `2 ** 10` | `1024` |

Shortcuts: `x += 1` means `x = x + 1` (also `-=`, `*=`, `/=`).

The `math` module adds functions such as `math.sqrt`, `math.exp`, `math.log` (natural log),
`math.factorial`, `math.comb` (n choose k) and the constants `math.pi` and `math.e`.
You'll use `math.comb` and `math.factorial` constantly in probability.

```python
import math

print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(math.sqrt(16), math.log(math.e), math.factorial(5), math.comb(5, 2))

count = 0
count += 1
count += 1
print(count)

# floats are approximations. This surprises everyone once:
print(0.1 + 0.2, 0.1 + 0.2 == 0.3, math.isclose(0.1 + 0.2, 0.3))
```

## 1.3 Comparisons and booleans

Comparisons produce a `bool`: `==` (equal), `!=` (not equal), `<`, `<=`, `>`, `>=`.
Combine booleans with `and`, `or`, `not`. Python also allows chains like `0 <= p <= 1`.

**Never compare floats with `==`.** Use `math.isclose(a, b)` (or `abs(a - b) < 1e-9`).

```python
p = 0.3
print(p > 0.5, p != 0.3, 0 <= p <= 1)
print(True and False, True or False, not True)
```

## 1.4 Strings and f-strings

Strings hold text: `'single'` or `"double"` quotes are the same thing.
An **f-string** (`f"..."`) evaluates whatever is inside `{}`.
Add a format spec after a colon: `{x:.3f}` gives 3 decimals and `{x:.1%}` a percentage.

```python
name = "Ben"
p_heads = 0.5
n = 10
print(f"Hello {name}! P(heads) = {p_heads}, n = {n}")
print(f"{math.pi:.3f}")        # 3.142
print(f"{0.12345:.1%}")         # 12.3%
print(f"{n=}")                  # handy for debugging: prints 'n=10'
print("ab" + "cd", "ha" * 3, len("hello"))
```

### Exercise 1.7: A formula with `**` and `math.exp`
The standard normal density is

$$\varphi(x) = \frac{1}{\sqrt{2\pi}}\, e^{-x^2/2}.$$

Compute `phi_0` $= \varphi(0)$ and `phi_1` $= \varphi(1)$ with `math.sqrt`, `math.pi` and `math.exp`.

<details><summary>Hint</summary>

`-x**2 / 2` already means $-(x^2)/2$ because `**` binds tighter than unary minus.
To be safe, write `-(x ** 2) / 2`.
</details>

```python
phi_0 = 1 / math.sqrt(2 * math.pi) * math.exp(-(0 ** 2) / 2)
phi_1 = 1 / math.sqrt(2 * math.pi) * math.exp(-(1 ** 2) / 2)
```

```python
assert math.isclose(phi_0, 0.3989422804014327)
assert math.isclose(phi_1, 0.24197072451914337)
print("Exercise 1.7 passed")
```
