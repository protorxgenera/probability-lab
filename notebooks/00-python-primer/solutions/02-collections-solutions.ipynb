## 2.1 Indexing and slicing (strings and lists)

Positions start at **0**. Negative indices count from the end: `-1` is the last element.

A **slice** `seq[start:stop:step]` takes elements from `start` **up to but not including** `stop`.
Any part can be left out: `seq[:3]` (first three), `seq[3:]` (from index 3 on), `seq[::-1]` (reversed).

```
 index:   0    1    2    3    4
 value:  'H'  'T'  'T'  'H'  'H'
 neg:    -5   -4   -3   -2   -1
```

````python
flips = "HTTHH"
print(flips[0], flips[-1])      # H H
print(flips[1:3])               # TT   (indices 1 and 2)
print(flips[:2], flips[2:])     # HT THH
print(flips[::2])               # HTH  (every second)
print(flips[::-1])              # HHTTH (reversed)
print(len(flips), flips.count("H"), "TT" in flips)
````

## 2.2 Lists: ordered and mutable

Lists hold any values, can be changed in place, and can grow.

| operation | effect |
|---|---|
| `xs.append(v)` | add `v` to the end |
| `xs.extend([a, b])` / `xs + ys` | add many / concatenate |
| `xs.insert(i, v)` | insert at position `i` |
| `xs.pop()` / `xs.pop(i)` | remove and return the last item / the item at `i` |
| `xs.remove(v)` | remove the first occurrence of `v` |
| `xs[i] = v` | replace one element |
| `sorted(xs)` / `xs.sort()` | new sorted list / sort in place |
| `sum(xs)`, `min(xs)`, `max(xs)`, `len(xs)` | built-in summaries |

````python
rolls = [3, 6, 1]
rolls.append(4)
rolls[0] = 5
print(rolls, len(rolls), sum(rolls), max(rolls))
print(sorted(rolls), rolls)            # sorted() does not change rolls
rolls.sort(reverse=True)
print(rolls)                           # .sort() does

# A list of lists, like a matrix
grid = [[1, 2, 3],
        [4, 5, 6]]
print(grid[1][2])                      # row 1, column 2 -> 6
````

### The aliasing trap

`b = a` does **not** copy a list. Both names point to the *same* list. Use `a.copy()` (or `list(a)`) for a real copy.
The same trap exists with NumPy arrays and pandas DataFrames.

````python
a = [1, 2, 3]
b = a            # same object
c = a.copy()     # new object
a.append(99)
print(a, b, c)
````

## 2.3 Tuples: ordered and immutable, great for unpacking

A tuple is like a list that can't change: `(1, 2, 3)`. You'll see them everywhere because functions
return several values as a tuple, which you **unpack**:

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
X_train, X_test, y_train, y_test = train_test_split(X, y)
fig, ax = plt.subplots()
```

Array shapes are tuples too: `matrix.shape` gives `(3, 2)`.

````python
point = (2.0, -1.5)
x, y = point
print(x, y)
first, *rest = [10, 20, 30, 40]    # star collects "the rest" into a list
print(first, rest)
# point[0] = 5   # uncomment: TypeError, tuples can't be modified
````

## 2.4 Dictionaries: key → value

A `dict` maps **keys** (usually strings) to **values**. Look up by key, not by position.

| operation | effect |
|---|---|
| `d[key]` | get (KeyError if missing) |
| `d.get(key, default)` | get, or `default` if missing |
| `d[key] = v` | add or overwrite |
| `key in d` | membership test on keys |
| `d.keys()`, `d.values()`, `d.items()` | views of keys, values, and (key, value) pairs |

````python
person = {"name": "John", "age": 30}
person["age"] = 31
person["city"] = "Bucharest"
print(person["name"], person.get("email", "no email"))
print(list(person.keys()), "age" in person)
for key, value in person.items():
    print(key, "->", value)
````

## 2.5 Sets: unique elements and set algebra

A `set` keeps only unique items and has no order. Set operations are the sample-space operations from probability:

| Python | math |
|---|---|
| `A \| B` | $A \cup B$ |
| `A & B` | $A \cap B$ |
| `A - B` | $A \setminus B$ |
| `x in A` | $x \in A$ |

````python
omega = {1, 2, 3, 4, 5, 6}         # one die roll
even = {2, 4, 6}
high = {4, 5, 6}
print(even | high, even & high, omega - even)
print(set([1, 1, 2, 2, 3]))         # removes duplicates
print(len(even & high) / len(omega))  # P(even and high) for a fair die
````

### Exercise 2.9: Dictionary views
Using the `pmf` dict below (outcome → probability):
- `outcomes`: a **list** of the keys
- `total_prob`: the sum of the values (should be 1)
- `most_likely`: the key with the largest probability. Look up how `max(pmf, key=pmf.get)` works.

<details><summary>Hint</summary>

`max(iterable, key=f)` returns the element `x` for which `f(x)` is largest.
Iterating a dict gives its keys, and `pmf.get(k)` gives the value for key `k`.

````python
pmf = {"rain": 0.3, "sun": 0.5, "snow": 0.2}
outcomes = list(pmf.keys())
total_prob = sum(pmf.values())
most_likely = max(pmf, key=pmf.get)
````

### Exercise 2.10: Events as sets
Roll one fair die. With `omega = {1, ..., 6}`:
- `A` = "roll is even", `B` = "roll is at least 4" (write both as sets)
- `A_or_B`, `A_and_B`, `not_A` (complement of A within omega)
- `p_A_or_B` = $|A \cup B| / |\Omega|$

````python
omega = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {4, 5, 6}
A_or_B = A | B
A_and_B = A & B
not_A = omega - A
p_A_or_B = len(A_or_B) / len(omega)
````

````python
assert A == {2, 4, 6} and B == {4, 5, 6}
assert A_or_B == {2, 4, 5, 6} and A_and_B == {4, 6} and not_A == {1, 3, 5}
assert abs(p_A_or_B - 4 / 6) < 1e-12
print("Exercise 2.10 passed")
````
