## 8.1 Two styles: `plt.*` and `fig, ax`

Matplotlib has two interfaces, and the MLMA lab uses both:

```python
# "pyplot" style: quick, acts on the "current" plot
plt.plot(x, y)
plt.title("...")
plt.show()

# object-oriented style: explicit, scales to many panels  <- use this one
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("...")
```

The names map almost one-to-one: `plt.title` ↔ `ax.set_title`, `plt.xlabel` ↔ `ax.set_xlabel`,
`plt.xlim` ↔ `ax.set_xlim`. **This primer uses `fig, ax`**, because with several subplots it's the only way that stays sane.

````python
x = np.linspace(0, 10, 200)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, np.sin(x), label="sin", color="tab:blue")
ax.plot(x, np.cos(x), label="cos", color="tab:red", linestyle="--")
ax.set_title("Two curves")
ax.set_xlabel("x")
ax.set_ylabel("value")
ax.legend()
plt.show()
````

## 8.2 The plot types you'll use most

| call | use it for |
|---|---|
| `ax.plot(x, y)` | functions, running averages, curves |
| `ax.scatter(x, y, s=..., alpha=...)` | point clouds, relationships between two variables |
| `ax.hist(data, bins=..., density=True)` | distribution of samples. `density=True` scales the bars so their total area is 1, so you can overlay a PDF |
| `ax.bar(categories, heights)` | PMFs, counts per category |
| `ax.step(x, y, where="post")` | CDFs |
| `ax.axhline(y)` / `ax.axvline(x)` | reference lines (true mean, threshold) |
| `ax.fill_between(x, y1, y2)` | shaded areas (probability as area under a density) |
| `ax.contour(X, Y, Z)` | 2-D functions and densities |

Common styling keywords: `color`, `linestyle` (`"-"`, `"--"`, `":"`), `linewidth`, `marker` (`"o"`, `"x"`), `alpha` (transparency), `label` (for the legend).

````python
rng = np.random.default_rng(0)
samples = rng.normal(0, 1, size=5000)
grid = np.linspace(-4, 4, 200)
pdf = np.exp(-grid ** 2 / 2) / np.sqrt(2 * np.pi)

fig, axes = plt.subplots(1, 3, figsize=(12, 3))     # 1 row, 3 columns: axes is an array of 3 Axes
axes[0].hist(samples, bins=40, density=True, alpha=0.6, label="samples")
axes[0].plot(grid, pdf, color="black", label="N(0,1) pdf")
axes[0].legend()
axes[0].set_title("Histogram + density")

axes[1].bar([1, 2, 3, 4, 5, 6], [1 / 6] * 6)
axes[1].set_title("PMF of a fair die")

axes[2].scatter(samples[:300], samples[:300] + rng.normal(0, 0.5, 300), s=8, alpha=0.5)
axes[2].set_title("Scatter")
fig.tight_layout()          # prevents overlapping labels
plt.show()
````

## 8.3 Subplot grids

`fig, axes = plt.subplots(nrows, ncols)` gives an array of axes: 1-D if one of the sizes is 1, 2-D otherwise (`axes[row, col]`).
`sharex=True` / `sharey=True` lines up the axes. Save a figure with `fig.savefig("name.png", dpi=150, bbox_inches="tight")`.
