## 3.1 `if` / `elif` / `else`

Python checks the conditions top to bottom and runs **only the first** block whose condition is true.

````python
x = 10
if x > 5:
    print(f"{x} is greater than 5")
elif x == 5:
    print(f"{x} is 5")
else:
    print(f"{x} is less than 5")

# One-line conditional expression ("ternary"):
label = "even" if x % 2 == 0 else "odd"
print(label)
````

## 3.2 `for` loops

`for item in collection:` runs the block once per item. Useful companions:

| tool | gives you | example |
|---|---|---|
| `range(n)` | `0, 1, ..., n-1` | `for i in range(5)` |
| `range(a, b, step)` | `a, a+step, ...` (stops before `b`) | `range(2, 11, 2)` → 2, 4, …, 10 |
| `enumerate(xs)` | `(index, item)` pairs | `for i, x in enumerate(xs)` |
| `zip(xs, ys)` | items side by side | `for x, y in zip(xs, ys)` |
| `d.items()` | `(key, value)` pairs | `for k, v in d.items()` |

The **accumulator pattern** is the most common loop you'll write: start from an empty/zero value and update it each iteration.

````python
for i in range(3):
    print("iteration", i)

names = ["Ana", "Bogdan", "Cris"]
scores = [0.9, 0.7, 0.8]
for i, name in enumerate(names):
    print(i, name)
for name, score in zip(names, scores):
    print(f"{name}: {score}")

# accumulator: sum of squares 1..5
total = 0
for k in range(1, 6):
    total += k ** 2
print(total)
````

## 3.3 `while`, `break`, `continue`

- `while condition:` repeats as long as the condition is true. Make sure something inside changes it!
- `break` exits the loop immediately. `continue` skips to the next iteration.

````python
count = 0
while count < 3:
    print("count =", count)
    count += 1

for k in range(10):
    if k % 2 == 1:
        continue      # skip odd numbers
    if k > 6:
        break         # stop entirely
    print(k)
````

## 3.4 Comprehensions

A **list comprehension** builds a list in one line:

```python
[expression for item in iterable if condition]
```

reads as *"the list of `expression` for each `item` in `iterable` (keeping only those where `condition`)"*. It's equivalent to:

```python
result = []
for item in iterable:
    if condition:
        result.append(expression)
```

There are also dict comprehensions `{k: v for ...}` and set comprehensions `{x for ...}`.

````python
squares = [k ** 2 for k in range(6)]
even_squares = [k ** 2 for k in range(6) if k % 2 == 0]
name_lengths = {name: len(name) for name in names}
print(squares, even_squares, name_lengths)
````

## 3.5 Random numbers in plain Python

The `random` module generates pseudo-random numbers. **Seeding** makes them reproducible: the same seed always gives the same sequence.
(Later you'll use NumPy's generator for this, but the idea is the same.)

- `random.random()`: a float uniformly in $[0, 1)$
- `random.randint(a, b)`: an integer in $\{a, \dots, b\}$ (both ends included!)
- `random.choice(seq)`: one random element

````python
import random

random.seed(0)
print(random.random(), random.randint(1, 6), random.choice(["H", "T"]))

# A Bernoulli(p) trial: "success" with probability p
p = 0.3
success = random.random() < p
print(success)
````
